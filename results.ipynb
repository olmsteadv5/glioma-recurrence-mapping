# Results

Run `./main.py 36 42 51` first. This just reads `outputs/results.json`.

```python
import json
import plotly.io as pio

from paths import RESULTS

# PyCharm does not render plotly's own mime type, so emit plain HTML instead
pio.renderers.default = "notebook_connected"

results = json.load(open(RESULTS))
sorted(results)
```

```python
for name in sorted(results):
    r = results[name]["regions_ml"]
    d = results[name].get("dice")
    line = f"{name}  WT {r['WT']}  TC {r['TC']}  ET {r['ET']}"
    if d:
        line += f"  dice {d['WT']} / {d['TC']} / {d['ET']}"
    print(line)
```

## Parcels the tumour sits in

```python
subject = "UPENN-GBM-00036_11"

a = results[subject]["atlas"]
print(f"{a['tumour_ml']} mL tumour, {a['in_parcels_ml']} mL in parcels")
for p in a["parcels"][:15]:
    print(f"{p['fraction']:>6.0%}  {p['long_name']:<42} {p['lobe']:<10} {p['tissue']}")
```
